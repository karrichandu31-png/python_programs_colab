```python
#Find Area of Circle
pi=3.14
radius=int(input("enter radius: "))
area=pi*radius*radius
print(area)
```

```python
#Find Perimeter of Rectangle
length=int(input("enter lenght:"))
breadth=int(input("enter breadth:"))
perimeter=2*(length+breadth)
print(perimeter)
```

```python
#Convert Days into Years and Months
days=int(input("enter no. of days:"))
years=days//365
remaining_days=days%365
months=remaining_days//30
print(f"{years} years {months} months")
```

```python
#Convert Temperature Celsius to Fahrenheit
Celsius=int(input("enter celsius:"))
Fahrenheit=(Celsius*(9/5))+32
print(f"{Fahrenheit:.2f} fahrenheit")
```

```python
#Find Simple Interest
principal=int(input("enter principal:"))
rate=int(input("enter rate:"))
time=int(input("enter years:"))
si=(principal*rate*time)/100
print(f"simple intrest={si}")
```

```python
#Find Compound Interest
principal=int(input("enter principal:"))
rate=int(input("enter rate:"))
time=int(input("enter years:"))
amount=principal*(1+rate/100)**time
ci=amount-principal
print(f"compound intrest={ci:.0f}")
```

```python
#Find Total Marks and Percentage
n=int(input("enter no. of subjects:"))
marks=[]
for i in range(n):
  m=int(input(f"enter marks of subject{i+1}:"))
  marks.append(m)
max_per_sub=int(input("enter maximum marks per subject:"))
total=sum(marks)
max_marks=n*max_per_sub
per=(total/max_marks)*100
print(f"total marks={total}")
print(f"percentage={per}%")
```

```python
#Convert Minutes into Hours
h=int(input("enter hours:"))
m=h/60
print(f"{m}hours")
```

```python
#Find Last Digit of Number Using Modulus
n=int(input("enter number:"))
l=n%10
print(f"last digit is={l}")
```

```python
#Find Square Root Using Exponent Operator
n=int(input("enter number:"))
s=n**(1/2)
print(f"square root={s:.0f}")
```
