```python
# 1. Increase student marks by bonus marks

marks = 75
bonus = 5

marks += bonus

print("Updated marks =", marks)
```

```python
# 2. Reduce product price after discount

price = 1000
discount = 200

price -= discount

print("Final price =", price)
```

```python
# 3. Multiply employee bonus by 3

bonus = 2000

bonus *= 3

print("Updated bonus =", bonus)
```

```python
# 4. Divide pizza slices among friends

pizza_slices = 12
friends = 4

pizza_slices /= friends

print("Slices per friend =", pizza_slices)
```

```python
# 5. Find remainder after division

num = 25

num %= 7

print("Remainder =", num)
```

```python
# 6. Create bank deposit program using +=

balance = 5000
deposit = 2000

balance += deposit

print("Updated balance =", balance)
```

```python
# 7. Create electricity bill reduction using -=

bill = 3500
discount = 500

bill -= discount

print("Reduced bill =", bill)
```

```python
# 8. Find cube using **=

num = 2

num **= 3

print("Cube =", num)
```

```python
# 9. Create score update system for game

score = 100

score += 50
score *= 2

print("Updated score =", score)
```

```python
# 10. Use all assignment operators in one program

a = 20

a += 5
print("After += :", a)

a -= 3
print("After -= :", a)

a *= 2
print("After *= :", a)

a /= 2
print("After /= :", a)

a %= 12
print("After %= :", a)

a **= 2
print("After **= :", a)

a //= 2
print("After //= :", a)
```
